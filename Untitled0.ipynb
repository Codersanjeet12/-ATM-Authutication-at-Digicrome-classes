```python
"""
main.py
-------
CLI entry point for the Enterprise Compliance Audit System.

Pipeline:
    logs (CSV/JSON/JSONL)
        -> log_ingestion (streaming)
        -> rule_engine (configurable YAML/JSON rules)
        -> tamper_evidence (SHA-256 hash-chained findings)
        -> report_generator (Excel + HTML, non-technical output)

Usage:
    python main.py --logs sample_data/sample_logs.csv \\
                    --rules rules/compliance_rules.yaml \\
                    --outdir output --title "Q3 Access Log Audit"
"""

from __future__ import annotations

import argparse
import sys
import time

from log_ingestion import ingest, LogIngestionError
from rule_engine import RuleEngine, RuleConfigError
from tamper_evidence import TamperEvidentLedger
from report_generator import build_excel_report, build_html_report


def run_audit(logs_path: str, rules_path: str, outdir: str, title: str) -> dict:
    print(f"[1/4] Loading rule set: {rules_path}")
    engine = RuleEngine.from_file(rules_path)
    print(f"      -> {len(engine.rules)} compliance rules loaded")

    print(f"[2/4] Streaming logs: {logs_path}")
    ledger = TamperEvidentLedger()

    start = time.perf_counter()
    total = 0
    for record in ingest(logs_path):
        total += 1
        for violation in engine.evaluate_record(record):
            ledger.append(violation)
        if total % 5000 == 0:
            print(f"      ...{total:,} records processed")
    elapsed = time.perf_counter() - start
    print(f"      -> {total:,} records scanned in {elapsed:.3f}s "
          f"({(total/elapsed if elapsed else 0):,.0f} records/sec)")
    print(f"      -> {len(ledger.chain):,} violations found")

    print("[3/4] Sealing audit trail (SHA-256 hash chain)")
    ok, problems = ledger.verify()
    print(f"      -> integrity check: {'PASSED' if ok else 'FAILED: ' + '; '.join(problems)}")
    print(f"      -> audit seal: {ledger.audit_seal()}")

    print(f"[4/4] Writing reports to: {outdir}")
    xlsx_path = build_excel_report(
        ledger, total, len(engine.rules), elapsed,
        f"{outdir}/compliance_audit_report.xlsx", audit_title=title,
    )
    html_path = build_html_report(
        ledger, total, len(engine.rules), elapsed,
        f"{outdir}/compliance_audit_report.html", audit_title=title,
    )
    print(f"      -> {xlsx_path}")
    print(f"      -> {html_path}")

    return {
        "records_scanned": total,
        "rules_applied": len(engine.rules),
        "violations_found": len(ledger.chain),
        "processing_seconds": elapsed,
        "integrity_ok": ok,
        "audit_seal": ledger.audit_seal(),
        "excel_report": xlsx_path,
        "html_report": html_path,
    }


def main():
    parser = argparse.ArgumentParser(description="Enterprise Compliance Audit System")
    parser.add_argument("--logs", required=True, help="Path to log file (.csv, .json, or .jsonl)")
    parser.add_argument("--rules", required=True, help="Path to rules file (.yaml or .json)")
    parser.add_argument("--outdir", default="output", help="Output directory for reports")
    parser.add_argument("--title", default="Enterprise Compliance Audit Report", help="Report title")
    args = parser.parse_args()

    try:
        summary = run_audit(args.logs, args.rules, args.outdir, args.title)
    except (LogIngestionError, RuleConfigError) as e:
        print(f"ERROR: {e}", file=sys.stderr)
        sys.exit(1)

    print("\n===== AUDIT COMPLETE =====")
    for k, v in summary.items():
        print(f"{k}: {v}")


if __name__ == "__main__":
    main()
```
