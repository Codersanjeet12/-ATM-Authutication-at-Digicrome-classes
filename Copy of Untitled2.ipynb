ATM Authentication

Create an ATM withdrawal program.
Rules:
If PIN is incorrect → "Invalid PIN"
If PIN is correct but withdrawal amount is greater than balance → "Insufficient Balance"
If amount is not a multiple of ₹100 → "Enter amount in multiples of 100"
Otherwise, process the withdrawal.

```python
# ATM Withdrawal Program

# preset PIN and balance
correct_pin = "1234"
balance = 5000

# user input
pin = input("Enter PIN: ")

if pin != correct_pin:
    print("Invalid PIN")
else:
    amount = int(input("Enter withdrawal amount: "))

    if amount > balance:
        print("Insufficient Balance")
    elif amount % 100 != 0:
        print("Enter amount in multiples of 100")
    else:
        balance -= amount
        print("Withdrawal successful!")
        print("Remaining Balance: ₹", balance)
```
